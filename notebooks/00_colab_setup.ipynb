```python
# Clone repository if not already available

import os
import subprocess
import sys
from pathlib import Path

runtime_root = Path.cwd().resolve()
repository_name = "minRF_interpolation"
candidates = [
    runtime_root / repository_name,
    runtime_root,
    Path("/content") / repository_name,
    Path("/content/minrf") / repository_name,
]
repo_root = next(
    (path.resolve() for path in candidates if (path / "api" / "flow_model_api.py").is_file()),
    None,
)

if repo_root is None:
    repo_root = Path("/content") / repository_name
    if repo_root.exists() and any(repo_root.iterdir()):
        raise FileExistsError(f"Cannot clone into non-empty directory: {repo_root}")
    subprocess.run(
        ["git", "clone", "https://github.com/Felinchen76/minRF_interpolation.git", str(repo_root)],
        check=True,
    )

if (repo_root / ".git").exists():
    subprocess.run(["git", "-C", str(repo_root), "pull", "--ff-only"], check=True)

os.environ.setdefault(
    "MINRF_RUNTIME_ROOT",
    str(runtime_root if runtime_root != repo_root else repo_root.parent),
)
os.chdir(repo_root)
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

print(f"Repository root: {repo_root}")
print(f"Runtime root: {os.environ['MINRF_RUNTIME_ROOT']}")
```
